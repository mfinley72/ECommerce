```python
import pandas as pd
import pyodbc

# 1. Define your local windows authentication connection parameters
server_name = r'DESKTOP-ANO09F8\SQLEXPRESS'  
database_name = 'EcommerceDB'

conn_str = (
    f"DRIVER={{SQL Server}};"
    f"SERVER={server_name};"
    f"DATABASE={database_name};"
    f"Trusted_Connection=yes;"  
)

# 2. Establish the connection gate
conn = pyodbc.connect(conn_str)

# 3. Write your SQL data extraction query string
query = "SELECT * FROM raw_ecommerce_data;"

# 4. Pull the database rows directly into a Pandas DataFrame
df = pd.read_sql(query, conn)

# 5. Close the database gateway
conn.close()

# Verify the data pulled through cleanly
print(f"Data successfully pulled from SQL Server! Total Rows: {len(df)}")
df.head()
```

```python
# 1. Remove canceled transactions (Invoice strings starting with 'C')
df = df[~df['InvoiceNo'].str.startswith('C', na=False)]

# 2. Strip rows with negative quantities or invalid unit prices
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

# 3. Create Total Sales feature logic
df['Total_Sales'] = df['Quantity'] * df['UnitPrice']

# 4. Extract precise time periods for dashboard trending metrics
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Year_Month'] = df['InvoiceDate'].dt.to_period('M').astype(str)
df['Day_of_Week'] = df['InvoiceDate'].dt.day_name()

# 5. Export clean data file for your dashboards
df.to_csv(r'C:\Users\Dell Precison\Documents\Python\Data Analysis\Ecommerce\cleaned_ecommerce_data.csv', index=False)
print("E-commerce data engineered and exported locally!")
```
